[문제1]

Python의 변수 할당 개념을 다른 언어와 비교해 설명해주세요


**python**에서 변수 선언 시 별도의 타입 지정이 필요하지 않으며 객체에 대한 참조(reference)를 저장하는 방식으로 작동하게 됩니다. 즉 변수는 특정 객체를 가리키는 라벨 역할을 하며 객체에 할당된 메모리 위치를 참조합니다.
```
x = 4
y = x
x = 7
```
이 코드에서 x에 4를 할당하면 4라는 객체가 생성되고 x는 그 객체를 참조합니다. 이어서 y = x를 실행하면 y는 x가 참조하는 같은 객체인 4를 참조하게 됩니다. 여기서 x = 7을 실행하게 되면 x는 새로운 객체인 7을 참조하게 되고 y는 여전히 4를 참조하고 있습니다.

C와 같은 언어에서는 변수가 실제 메모리 주소에 직접 매핑되기 때문에 변수를 선언할 때 명시적으로 데이터 타입을 지정하면 데이터 타입에 맞는 메모리 공간이 할당되게 됩니다.
```
int x = 4;
int y = x;
x = 10;
```
이 코드에서 x = 4는 메모리 상의 특정 위치에 4라는 값을 직접 저장합니다. y = x를 실행하면 y도 동일한 값 4를 저장하는 새로운 메모리 공간을 가지게 되며 x와 y는 별도의 메모리 공간을 차지하게 됩니다. x = 10을 실행하면, x의 값만 10으로 바뀌고 y의 값은 여전히 4로 유지됩니다. python과 달리 c언어는 참조가 아닌 값 자체가 복사됩니다.

python의 변수 할당은 객체 지향적이며 메모리 관리를 자동으로 해주는 방식이지만 c언어나 다른 언어들은 메모리 관리를 좀 더 명확하게 하도록 설계되어 있습니다.
